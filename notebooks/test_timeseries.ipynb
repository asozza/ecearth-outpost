```python
import numpy as np
import xarray as xr
import matplotlib.pyplot as plt
```

```python
from ecpost.core.graphics.timeseries import timeseries
```

```python
timeseries(expname='cs00', startyear=1990, endyear=4999, varname='thetao', avetype='moving', label='COLD-START')
timeseries(expname='cw00', startyear=1990, endyear=4999, varname='thetao', avetype='moving', label='WOA13')
plt.legend(
    bbox_to_anchor=(0.98, 0.98),  # x, y coordinates for legend placement
    loc='upper right',         # Location of the legend relative to bbox_to_anchor
    borderaxespad=0           # Padding between the legend and the plot
)
plt.title('Coupled Mode')
```

```python
timeseries(expname='os00', startyear=1990, endyear=4999, varname='thetao', avetype='moving', label='COLD-START')
timeseries(expname='ow00', startyear=1990, endyear=4999, varname='thetao', avetype='moving', label='WOA13')
plt.legend(
    bbox_to_anchor=(0.98, 0.98),  # x, y coordinates for legend placement
    loc='upper right',         # Location of the legend relative to bbox_to_anchor
    borderaxespad=0           # Padding between the legend and the plot
)
plt.title("OMIP Mode")
```
