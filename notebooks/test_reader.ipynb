```python
import matplotlib.pyplot as plt
import numpy as np

ds = netCDF4.Dataset(filename)

var = ds.variables["thetao"]   # cambia nome variabile
data = var[0, 0, :, :]         # time=0, depth=0

plt.figure()
plt.imshow(data, origin="lower")
plt.colorbar()
plt.title("thetao t=0 z=0")
plt.savefig("thetao.png")
plt.close()
```

```python
da = xr.DataArray(
    ds.variables["thetao"][:],
    dims=("time", "depth", "y", "x"),
    name="thetao"
)
da.isel(time=0, depth=0).plot()
```
